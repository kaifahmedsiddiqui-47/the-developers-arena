this heatmap shows how many null values are there in the dataset

there are too many null values in the dataset,
        the solution is that we can either drop them or we can get the mean value and replace the null values
        the mean values to get the accurate calculation in further eda
        here we will replace the null values with the mean values


```python
null_count = df[df["house_size"].isna()]
print("Number of null values in house_size:", null_count)
```

```python
df['house_size'] = df['house_size'].fillna(0)  
```

```python
df["bed"].mean()
```

```python

df['bed'] = df['bed'].fillna(3.275)
```

```python
df["bath"].mean()
```

```python
df['bath']=df["bath"].fillna(2.496440300452989)
```

```python
df["prev_sold_date"] = pd.to_datetime(df["prev_sold_date"], errors='coerce')
mean_date = df["prev_sold_date"].mean()
print(mean_date)
```

```python
df["prev_sold_date"]=df["prev_sold_date"].fillna(2017-8-16)
```

```python
df["acre_lot"].mean()
```

```python
df["acre_lot"]=df["acre_lot"].fillna(15.223027446965554)
```

```python
df["brokered_by"].mean()
```

```python
df["brokered_by"]=df["brokered_by"].fillna(52939.89328662749)
```

```python
df["street"].mean()
```

```python
df["street"]=df["street"].fillna(1012324.9206654341)
```

```python
sns.heatmap(df.isnull(),yticklabels=False,cbar=False,cmap='viridis')
```

now all the null values have been handeled 
this data set is now called as cleaned data set

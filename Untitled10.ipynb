```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
```

```python
df=pd.read_csv("D:\data  science\pandas\car_sales_data.csv")
```

```python
df.groupby('Manufacturer')['Price'].mean()
```

```python
sns.catplot(x='Manufacturer', y='Price', kind='bar', data=df) 
```

```python
x=df["Year of manufacture"]
y=df["Price"]
plt.bar(x,y)
plt.xlabel("year of manufacture")
plt.ylabel("price")
plt.show
```

```python
sns.violinplot(
   x=df["Fuel type"],
   y=df["Price"]
)
plt.show
```

```python
sns.boxplot(x="Manufacturer", y="Engine size",data=df)
plt.show()
```

```python
plt.figure(figsize=(10,10))
sns.heatmap(df.select_dtypes(include='number').corr(), annot=True, cmap="RdYlGn")
plt.show()
```

```python
sns.pairplot(df[["Engine size", "Year of manufacture", "Mileage", "Price"]])
plt.show()
```

```python
fig = px.scatter(
    df, x="Year of manufacture", y="Price",
    color="Fuel type",
    animation_frame="Fuel type", 
   # animation_group="Manufacturer"

)
fig.show()
```

```python
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots

fig_bar = px.bar(df, x="Manufacturer", y="Price", title="Manufacturer vs Price")
```

```python
fig_year = px.bar(df, x="Year of manufacture", y="Price", title="Year vs Price")
```

```python
fig_violin = px.violin(df, x="Fuel type", y="Price", box=True, points="all", title="Fuel type vs Price")
```

```python
fig_box = px.box(df, x="Manufacturer", y="Engine size", title="Manufacturer vs Engine Size")
```

```python
import plotly.figure_factory as ff

corr = df.select_dtypes(include='number').corr()
fig_heatmap = ff.create_annotated_heatmap(
    z=corr.values,
    x=list(corr.columns),
    y=list(corr.index),
    colorscale="RdYlGn"
)
fig_heatmap.update_layout(title="Correlation Heatmap")
```

```python
fig_pair = px.scatter_matrix(
    df[["Engine size", "Year of manufacture", "Mileage", "Price"]],
    dimensions=["Engine size", "Year of manufacture", "Mileage", "Price"],
    color="Price"
)
fig_pair.show()
```

```python
fig_anim = px.scatter(
    df, x="Year of manufacture", y="Price",
    color="Fuel type",
    animation_frame="Fuel type",
    title="Animated Scatter by Fuel Type"
)
```

```python
from plotly.subplots import make_subplots

fig = make_subplots(
    rows=3, cols=2,
    subplot_titles=("Bar: Manufacturer vs Price", "Bar: Year vs Price",
                    "Violin: Fuel vs Price", "Box: Manufacturer vs Engine Size",
                    "Heatmap: Correlation")
)

for trace in fig_bar.data:
    fig.add_trace(trace, row=1, col=1)

for trace in fig_year.data:
    fig.add_trace(trace, row=1, col=2)

for trace in fig_violin.data:
    fig.add_trace(trace, row=2, col=1)

for trace in fig_box.data:
    fig.add_trace(trace, row=2, col=2)

for trace in fig_heatmap.data:
    fig.add_trace(trace, row=3, col=1)


fig.update_layout(height=1200, width=1200, title_text="Combined Dashboard of Visualizations")
fig.show()
```

```python
fig.write_html("dashboard.html")
```

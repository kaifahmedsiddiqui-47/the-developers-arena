```python
merged['Date'] = pd.to_datetime(merged['Date'], errors='coerce')
```

```python
jan_sales = merged[(merged['Date'] >= '2024-01-01') & (merged['Date'] <= '2024-01-31')]
jan_sales = merged[(merged['Date'].dt.month == 1) & (merged['Date'].dt.year == 2024)]
feb_sales = merged[(merged['Date'] >= '2024-02-01') & (merged['Date'] <= '2024-02-28')]
feb_sales = merged[(merged['Date'].dt.month == 1) & (merged['Date'].dt.year == 2024)]
mar_sales = merged[(merged['Date'] >= '2024-03-01') & (merged['Date'] <= '2024-03-31')]
mar_sales = merged[(merged['Date'].dt.month == 1) & (merged['Date'].dt.year == 2024)]
apr_sales = merged[(merged['Date'] >= '2024-04-01') & (merged['Date'] <= '2024-04-30')]
apr_sales = merged[(merged['Date'].dt.month == 1) & (merged['Date'].dt.year == 2024)]
```

```python
jan_sales.head()
```

```python
feb_sales.head()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 8)) 

# January
axes[0,0].plot(jan_sales['Date'], jan_sales['Total_Sales'], marker='o')
axes[0,0].set_title("January Sales")
axes[0,0].set_xlabel("Date")
axes[0,0].set_ylabel("Sales")

# February
axes[0,1].plot(feb_sales['Date'], feb_sales['Total_Sales'], marker='o', color='orange')
axes[0,1].set_title("February Sales")
axes[0,1].set_xlabel("Date")
axes[0,1].set_ylabel("Sales")

# March
axes[1,0].plot(mar_sales['Date'], mar_sales['Total_Sales'], marker='o', color='green')
axes[1,0].set_title("March Sales")
axes[1,0].set_xlabel("Date")
axes[1,0].set_ylabel("Sales")

# april
axes[1,1].plot(apr_sales['Date'], apr_sales['Total_Sales'], marker='o', color='red')
axes[1,1].set_title("April Sales")
axes[1,1].set_xlabel("Date")
axes[1,1].set_ylabel("Sales")

plt.tight_layout()
plt.show()
```

This subplots show the total sales of every month with respect to dates and the product prices

```python
pivot_multi = pd.pivot_table(
    merged,
    values='Total_Sales',
    index='Region',
    columns='Product',
    aggfunc=['sum','mean','count'],
    fill_value=0
)

print(pivot_multi)
```

```python
pivot_customer = pd.pivot_table(
    merged,
    values='Total_Sales',
    index='Customer_ID',
    aggfunc='sum'
).sort_values('Total_Sales', ascending=False)

print(pivot_customer.head(10))
```

```python
pivot_basic = pd.pivot_table(
    merged,
    values='Total_Sales',
    index='Region',
    columns='Product',
    aggfunc='sum',
    fill_value=0
)

print(pivot_basic)
```

```python
merged['Month'] = merged['Date'].dt.to_period('M')

pivot_month = pd.pivot_table(
    merged,
    values='Total_Sales',
    index='Month',
    columns='Product',
    aggfunc='sum',
    fill_value=0
)

print(pivot_month)
```

```python


sns.heatmap(pivot_month, annot=True, fmt=".0f", cmap="Blues")
plt.title("Monthly Sales by Product")
plt.show()
```

```python
import plotly.express as px

# Monthly Sales Trend
merged['Month'] = merged['Date'].dt.to_period('M').astype(str)
monthly_sales = merged.groupby('Month')['Total_Sales'].sum().reset_index()
fig1 = px.line(monthly_sales, x='Month', y='Total_Sales', title="Monthly Sales Trend")

# Best-Selling Products
product_sales = merged.groupby('Product')['Total_Sales'].sum().reset_index()
fig2 = px.bar(product_sales, x='Product', y='Total_Sales', title="Best-Selling Products")

# Regional Sales Distribution
region_sales = merged.groupby('Region')['Total_Sales'].sum().reset_index()
fig3 = px.pie(region_sales, names='Region', values='Total_Sales', title="Regional Sales Distribution")

# Churn by Payment Method
pivot_churn = merged.pivot_table(values='Customer_ID', index='PaymentMethod', columns='Churn', aggfunc='count', fill_value=0).reset_index()
fig4 = px.bar(pivot_churn, x='PaymentMethod', y=[0,1], title="Churn by Payment Method", barmode='stack')
```

```python
from dash import Dash, html, dcc

app = Dash(__name__)

app.layout = html.Div(children=[
    html.H1("Sales & Customer Dashboard"),

    dcc.Graph(figure=fig1),
    dcc.Graph(figure=fig2),
    dcc.Graph(figure=fig3),
    dcc.Graph(figure=fig4),
])

if __name__ == '__main__':
    app.run(debug=True)
```

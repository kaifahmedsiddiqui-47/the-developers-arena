```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("D:\data  science\pandas\customer_churn.csv")
```

```python
df
```

```python
df.info()
```

```python
features = ['Tenure', 'MonthlyCharges', 'TotalCharges', 'SeniorCitizen']
X = df[features]

```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

```

```python
inertia = []
K = range(1, 11)
for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)
```

```python
plt.plot(K, inertia, 'bo-')
plt.xlabel('Number of clusters (k)')
plt.ylabel('Inertia')
plt.title('Elbow Method for Optimal k')
plt.show()
```

```python
 
optimal_k = 3
kmeans = KMeans(n_clusters=optimal_k, random_state=42)
df['Cluster'] = kmeans.fit_predict(X_scaled)


cluster_summary = df.groupby('Cluster')[['Tenure','MonthlyCharges','TotalCharges','SeniorCitizen']].mean()
print(cluster_summary)
```

➡️ These are long-term, high-spending younger customers. Likely Premium Spenders who are loyal and valuable.

```python
from sklearn.cluster import AgglomerativeClustering
from sklearn.metrics import silhouette_score
import scipy.cluster.hierarchy as sch
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10, 6))
dendrogram = sch.dendrogram(sch.linkage(X_scaled, method='ward'))
plt.title("Hierarchical Clustering Dendrogram")
plt.xlabel("Customers")
plt.ylabel("Euclidean distances")
plt.show()
```

```python

hc = AgglomerativeClustering(n_clusters=3, linkage='ward', metric='euclidean')
df['HC_Cluster'] = hc.fit_predict(X_scaled)
```

```python
silhouette_hc = silhouette_score(X_scaled, df['HC_Cluster'])
print("Hierarchical Silhouette Score:", silhouette_hc)
```

```python
from sklearn.cluster import DBSCAN

dbscan = DBSCAN(eps=0.8, min_samples=5)  # tune eps & min_samples
df['DBSCAN_Cluster'] = dbscan.fit_predict(X_scaled)

# Evaluate (ignore -1 which is noise)
valid_clusters = df[df['DBSCAN_Cluster'] != -1]
silhouette_db = silhouette_score(valid_clusters[features], valid_clusters['DBSCAN_Cluster'])
print("DBSCAN Silhouette Score:", silhouette_db)
```

```python
from sklearn.metrics import davies_bouldin_score

print("KMeans DB Index:", davies_bouldin_score(X_scaled, df['Cluster']))
print("HC DB Index:", davies_bouldin_score(X_scaled, df['HC_Cluster']))
print("DBSCAN DB Index:", davies_bouldin_score(valid_clusters[features], valid_clusters['DBSCAN_Cluster']))
```

Day 2 Insights
- K-Means (Day 1): Strong elbow at 3 clusters, interpretable customer groups.
- Hierarchical: Silhouette = 0.224 → weak separation, clusters overlap.
- DBSCAN: Silhouette ≈ 0 → no useful clusters, mostly noise.
👉 Conclusion: K-Means with 3 clusters is the most stable and business-friendly method for your dataset. Hierarchical confirms some grouping but weaker, DBSCAN isn’t suitable here.

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score
```

```python
features = ['Tenure','MonthlyCharges','TotalCharges','SeniorCitizen']
target = 'Churn'

results = {}

for cluster_id in df['Cluster'].unique():
    # Subset data for this cluster
    cluster_data = df[df['Cluster'] == cluster_id]
    X = cluster_data[features]
    y = cluster_data[target]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
    
    
rf = RandomForestClassifier(random_state=42)
rf.fit(X_train, y_train)
y_pred = rf.predict(X_test)
y_prob = rf.predict_proba(X_test)[:,1]
```

```python
results[cluster_id] = {
        'Accuracy': accuracy_score(y_test, y_pred),
        'Precision': precision_score(y_test, y_pred),
        'Recall': recall_score(y_test, y_pred),
        'F1': f1_score(y_test, y_pred),
        'ROC-AUC': roc_auc_score(y_test, y_prob)
    }

results
```

Cluster 0 Model Performance
- Accuracy: 92.5% → very strong overall correctness.
- Precision: 85.7% → when the model predicts churn, it’s right most of the time.
- Recall: 60% → but it misses quite a few actual churners.
- F1-score: 0.71 → balanced but not perfect.
- ROC-AUC: 0.99 → excellent ability to separate churn vs. non-churn.


 Day 4 Deliverables Recap
- Random Forest model trained for Cluster 0 (Premium Spenders).
- Performance metrics calculated.
- Business insight: The model is strong overall, but recall needs improvement (we’ll fix this in Day 6 hyperparameter tuning

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score

features = ['Tenure','MonthlyCharges','TotalCharges','SeniorCitizen']
target = 'Churn'

results = {}

```

```python
for cluster_id in df['Cluster'].unique():
    cluster_data = df[df['Cluster'] == cluster_id]
    X = cluster_data[features]
    y = cluster_data[target]
    
    # Skip clusters with only one class
    if len(y.unique()) < 2:
        print(f"Cluster {cluster_id} has only one class in target. Skipping ROC-AUC.")
        continue
    
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
    
    rf = RandomForestClassifier(random_state=42)
    rf.fit(X_train, y_train)
    y_pred = rf.predict(X_test)
    y_prob = rf.predict_proba(X_test)[:,1]
    
    results[cluster_id] = {
        'Accuracy': accuracy_score(y_test, y_pred),
        'Precision': precision_score(y_test, y_pred),
        'Recall': recall_score(y_test, y_pred),
        'F1': f1_score(y_test, y_pred),
        'ROC-AUC': roc_auc_score(y_test, y_prob)
    }

results
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestClassifier

# Parameter grid
param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [None, 5, 10, 20],
    'min_samples_split': [2, 5, 10],
    'class_weight': [None, 'balanced']
}
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score

tuned_results = {}

for cluster_id in df['Cluster'].unique():
    cluster_data = df[df['Cluster'] == cluster_id]
    X = cluster_data[features]
    y = cluster_data[target]
    
    # Skip if only one class
    if len(y.unique()) < 2:
        print(f"Cluster {cluster_id} has only one class. Skipping tuning.")
        continue
    
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
    
    rf = RandomForestClassifier(random_state=42)
    grid_search = GridSearchCV(rf, param_grid, cv=3, scoring='f1', n_jobs=-1)
    grid_search.fit(X_train, y_train)
    
    best_model = grid_search.best_estimator_
    y_pred = best_model.predict(X_test)
    y_prob = best_model.predict_proba(X_test)[:,1]
    
    tuned_results[cluster_id] = {
        'Best Params': grid_search.best_params_,
        'Accuracy': accuracy_score(y_test, y_pred),
        'Precision': precision_score(y_test, y_pred),
        'Recall': recall_score(y_test, y_pred),
        'F1': f1_score(y_test, y_pred),
        'ROC-AUC': roc_auc_score(y_test, y_prob)
    }

tuned_results
```
